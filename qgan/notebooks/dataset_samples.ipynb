# Dataset samples

Show every sample for the `base`, `ang`, or `amp` experiment preset. `base` uses the same full-basis probability plot as the results analysis output panels, including zero-probability states. `ang` and `amp` show gradient images with the pixel counts used by their respective encodings.


## Setup

Run with the project notebook environment (`pip install -e "qgan[results]"`). The path cell works from the repository root or `qgan/notebooks`. Generation and plotting code lives in `qgan_v2.visualization.datasets`.


```python
from pathlib import Path
import sys

cwd = Path.cwd().resolve()
qgan_dir = next(
    (
        candidate
        for root in (cwd, *cwd.parents)
        for candidate in (root, root / "qgan")
        if (candidate / "src" / "qgan_v2").is_dir()
    ),
    None,
)
if qgan_dir is None:
    raise RuntimeError("Could not find the qgan project directory.")

src_path = qgan_dir / "src"
if str(src_path) not in sys.path:
    sys.path.insert(0, str(src_path))

from qgan_v2.visualization import show_dataset_samples
```

## Choose preset and PNG export

Set `PRESET` to `"base"`, `"ang"`, or `"amp"`. Set `QUBIT_COUNTS` to any combination of 4, 8, and 16. All samples for each selected size are shown. Set `CREATE_PNG = True` to save one 300 dpi PNG per size in `FIGURE_DIR`.


```python
CREATE_PNG = False
FIGURE_DIR = qgan_dir / "figures" / "datasets"
```

```python
PRESET = "base"
QUBIT_COUNTS = (4, 8)
CREATE_PNG = True

figures = show_dataset_samples(
    PRESET,
    QUBIT_COUNTS,
    save_png=CREATE_PNG,
    figure_dir=FIGURE_DIR,
)
```

```python
PRESET = "ang"
QUBIT_COUNTS = (4, 8, 16)

figures = show_dataset_samples(
    PRESET,
    QUBIT_COUNTS,
    save_png=CREATE_PNG,
    figure_dir=FIGURE_DIR,
)
```

```python
PRESET = "amp"
QUBIT_COUNTS = (4, 8)

figures = show_dataset_samples(
    PRESET,
    QUBIT_COUNTS,
    save_png=CREATE_PNG,
    figure_dir=FIGURE_DIR,
)
```
